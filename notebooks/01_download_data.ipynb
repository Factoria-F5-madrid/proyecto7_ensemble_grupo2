```python
import os
import zipfile
import gdown
```

```python
# 📁 Definimos las rutas de trabajo
BASE_DIR = os.path.abspath(os.path.join(os.getcwd(), ".."))  # sube un nivel (desde /notebooks a /backend)
DATA_DIR = os.path.join(BASE_DIR, "data")                    # backend/data
MODELS_DIR = os.path.join(BASE_DIR, "backend/models")
RESULTS_DIR = os.path.join(BASE_DIR, "backend/results") 
ZIP_PATH = os.path.join(DATA_DIR, "desayuno_preprocessed_archive.zip")
EXTRACT_DIR = os.path.join(DATA_DIR, "desayuno_preprocessed")

print("📂 Directorios configurados:")
print("BASE_DIR:", BASE_DIR)
print("DATA_DIR:", DATA_DIR)
print("RESULTS_DIR:", RESULTS_DIR)
print("ZIP_PATH:", ZIP_PATH)
print("EXTRACT_DIR:", EXTRACT_DIR)

# Crear la carpeta data/ si no existe
os.makedirs(DATA_DIR, exist_ok=True)

# Crear la carpeta results/ si no existe
os.makedirs(RESULTS_DIR, exist_ok=True)

# Crear la carpeta models/ si no existe
os.makedirs(MODELS_DIR, exist_ok=True)
```

```python
# 🔗 URL del dataset limpio (Google Drive)
DRIVE_URL = "https://drive.google.com/uc?id=1ReMjp7GNUz1cm2o-V8nDDdtAO9shnUtE"

print("📥 Descargando dataset desde Google Drive...")
gdown.download(DRIVE_URL, ZIP_PATH, quiet=False)

# Comprobamos tamaño mínimo (para detectar descargas erróneas)
if os.path.getsize(ZIP_PATH) < 1_000_000:
    raise ValueError("❌ Archivo descargado demasiado pequeño. Verifica el enlace o los permisos de Drive.")

print(f"✅ Archivo descargado correctamente en: {ZIP_PATH}")
```

```python
# 🗂️  Extraemos el contenido del ZIP descargado
print("🗂️  Extrayendo contenido...")
os.makedirs(EXTRACT_DIR, exist_ok=True)

with zipfile.ZipFile(ZIP_PATH, 'r') as zip_ref:
    zip_ref.extractall(EXTRACT_DIR)

print(f"✅ Dataset descomprimido en: {EXTRACT_DIR}")
```

```python
# 🔍 Mostramos algunos archivos para confirmar la extracción
print("\n📁 Contenido del dataset:")
files = os.listdir(EXTRACT_DIR)
for f in files[:15]:
    print(" -", f)

print(f"\nTotal de archivos: {len(files)}")
```
